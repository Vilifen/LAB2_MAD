```python
## ПУНКТ 4
# В зависимости от типа переменных выбрать линейную модель регрессии или пуассоновскую регрессию.

# В качестве основной модели выбрана линейная регрессия, поскольку зависимая переменная price (стоимость недвижимости) 
# представляет собой непрерывную количественную характеристику, принимающую денежные значения в широком диапазоне. 
# Линейная регрессия позволяет оценить прямое влияние каждого фактора (sqft_living, bedrooms, waterfront) на изменение стоимости 
# и имеет наглядную экономическую интерпретацию коэффициентов.
# Пуассоновская регрессия в основном применяется для моделирования дискретных счетных данных (количества событий), 
# поэтому для непрерывной целевой переменной price с высокой вариативностью она является менее приоритетным выбором. 
# В рамках лабораторной работы пуассоновская модель строится дополнительно для сравнения качества её аппроксимации с линейной и нелинейной моделями.
```

```python
## ПУНКТ 5
# Построить линейную регрессионную модель с фиктивными переменными для выбранных признаков.


df_model = df[['price', 'sqft_living', 'bedrooms', 'waterfront']].copy()

# Проверяем наличие пропусков и приводим waterfront к целочисленному типу (если требуется)
df_model['waterfront'] = df_model['waterfront'].astype(int)

print(df_model[['price', 'sqft_living', 'bedrooms', 'waterfront']].head(15))

# Регрессия методом наименьших квадратов
y = df_model['price']
X = df_model[['sqft_living', 'bedrooms', 'waterfront']]

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())
```

```python
## ПУНКТ 6
# Вывести график остатков. Оценить постоянство среднего и дисперсии.
predicted = model.fittedvalues
residuals = model.resid

plt.figure(figsize=(10, 6))
plt.scatter(predicted, residuals, alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel('Предсказанная цена')
plt.ylabel('Остатки')
plt.title('График остатков относительно предсказанных значений')
plt.grid(alpha=0.3)
plt.show()
```

```python
## ПУНКТ 7
# Построить гистограмму стандартизированных остатков.
standardized_residuals = model.get_influence().resid_studentized_internal

n = len(standardized_residuals)
k = int(np.ceil(1 + 3.322 * np.log10(n)))

minimum = standardized_residuals.min()
maximum = standardized_residuals.max()

bins = np.linspace(minimum, maximum, k + 1)
h = (maximum - minimum) / k

print('n =', n)
print('k =', k)
print('min =', minimum)
print('max =', maximum)
print('h =', h)

print('\nИнтервалы:')
for i in range(k):
    print(f'{i + 1}: {bins[i]:.2f} — {bins[i + 1]:.2f}')

plt.figure(figsize=(10, 6))
plt.hist(standardized_residuals, bins=bins, edgecolor='black', alpha=0.7)
plt.xticks(bins, [f'{x:.1f}' for x in bins], rotation=45)
plt.xlabel('Стандартизированные остатки')
plt.ylabel('Частота')
plt.title('Гистограмма стандартизированных остатков')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
## ПУНКТ 9
# По значениям коэффициента детерминации, графику и гистограмме остатков оценить качество построенной модели.

# По значениям коэффициента детерминации построенная линейная регрессионная модель обладает средней объясняющей 
# способностью: R2 =0,526, скорректированный коэффициент детерминации Radj^2=0,525. Модель объясняет около 52,6% 
# вариации стоимости объектов недвижимости, что является вполне приемлемым результатом для пространственных и рыночных данных.

# Анализ графика остатков показывает, что точки в среднем располагаются вокруг нулевой линии (среднее остатков равно нулю), 
# однако с увеличением предсказанной цены наблюдается заметное увеличение разброса остатков (эффект гетероскедастичности). 
# Модель сильнее ошибается при оценке дорогих объектов недвижимости.

# Гистограмма стандартизированных остатков показывает, что основная их часть сосредоточена около нуля (преимущественно 
# в диапазоне от −0,8 до 0,2), однако распределение имеет выраженную правую асимметрию и содержит отдельные сильно отклоняющиеся 
# наблюдения (выбросы), выходящие за пределы диапазона ±3 (максимальное значение достигает 7,71). Это свидетельствует о 
# существенном отклонении распределения остатков от нормального закона.
```

```python
## ПУНКТ 10
# Оценить значимость построенной модели.
from scipy.stats import f

alpha = 0.05
df1 = 3  # количество факторов (sqft_living, bedrooms, waterfront)
df2 = n - df1 - 1  # число степеней свободы остатков

F_crit = f.ppf(1 - alpha, df1, df2)
print('F_крит =', F_crit)

# Значение эмпирического критерия Фишера F=553,5 (получено из сводной таблицы результатов регрессии) значительно превышает критическое 
# значение Fкрит=2,61 (при уровне значимости α=0,05 и числе степеней свободы df1=3, df2=1496). Так как F>Fкрит(553,5>2,61), 
# мы делаем вывод, что в целом уравнение регрессии является статистически значимым. 
# Следовательно, построенная модель имеет надежную статистическую связь между зависимой переменной (price) и совокупностью 
# включённых в нее факторов (sqft_living, bedrooms, waterfront).
```

```python
## ПУНКТ 12
# Построить нелинейную регрессионную модель с фиктивными переменными (повторить пункты 6-10). Можно использовать
# библиотечные функции для линейной свертки нелинейных элементов, например, y=aox3+a1x2*logx.

# Построим нелинейную регрессионную модель вида y=aox3+a1x2*logx
df_log = df[['price', 'sqft_living', 'bedrooms', 'waterfront']].copy()

# Формируем нелинейные трансформации x^3 и x^2 * log(x)
df_log['sqft_living_3'] = df_log['sqft_living'] ** 3
df_log['sqft_living_2_log'] = (
    df_log['sqft_living'] ** 2 *
    np.log(df_log['sqft_living'])
)

df_log['bedrooms_3'] = df_log['bedrooms'] ** 3
df_log['bedrooms_2_log'] = (
    df_log['bedrooms'] ** 2 *
    np.log(df_log['bedrooms'])
)

y = df_log['price']

X = df_log[
    [
        'sqft_living_3',
        'sqft_living_2_log',
        'bedrooms_3',
        'bedrooms_2_log',
        'waterfront'
    ]
]

X = sm.add_constant(X)

log_poly_model = sm.OLS(y, X).fit()

print(log_poly_model.summary())

# График остатков
fitted = log_poly_model.fittedvalues
residuals = log_poly_model.resid

plt.figure(figsize=(9, 6))

plt.scatter(fitted, residuals, alpha=0.5)

plt.axhline(0, color='red', linestyle='--')

plt.xlabel('Предсказанные значения цены')
plt.ylabel('Остатки')
plt.title('График остатков нелинейной регрессионной модели')

plt.grid(True, alpha=0.3)
plt.show()

# Гистограмма стандартизированных остатков
standardized_residuals = log_poly_model.get_influence().resid_studentized_internal

n = len(standardized_residuals)
k = int(np.ceil(1 + 3.322 * np.log10(n)))

xmin = standardized_residuals.min()
xmax = standardized_residuals.max()
bins = np.linspace(xmin, xmax, k + 1)

plt.figure(figsize=(9, 6))

plt.hist(
    standardized_residuals,
    bins=bins,
    edgecolor='black'
)

plt.xlabel('Стандартизированные остатки')
plt.ylabel('Частота')
plt.title('Гистограмма стандартизированных остатков нелинейной регрессионной модели')

plt.xticks(
    bins,
    [f'{x:.1f}' for x in bins],
    rotation=45
)

plt.grid(True, alpha=0.3)
plt.show()

# Уравнение регрессии
# price=286500−0.000005228⋅sqft_living^3+0.0086⋅sqft_living^2⋅ln(sqft_living)+296.0079⋅bedrooms^3−2317.9974⋅bedrooms^2⋅ln(bedrooms)+607200⋅waterfront

# По значениям коэффициента детерминации, графику и гистограмме остатков проведем оценку качества построенной нелинейной регрессионной модели

# По значениям коэффициента детерминации построенная нелинейная регрессионная модель обладает средней объясняющей способностью: 
# R2=0,543, скорректированный коэффициент детерминации Radj^2=0,541. Модель объясняет около 54,3% вариации стоимости недвижимости.

# По графику остатков можно отметить, что остатки в целом располагаются вокруг нулевого уровня (линии y=0), и выраженного 
# систематического тренда не наблюдается. Однако с увеличением предсказанного значения цены наблюдается явное увеличение разброса остатков, 
# образующее воронкообразную форму. Это свидетельствует о том, что дисперсия ошибок не является постоянной, то есть присутствует гетероскедастичность.

# Гистограмма стандартизированных остатков показывает, что основная их часть сосредоточена вблизи нулевого значения (преимущественно в 
# интервале от −0,7 до 0,2). Вместе с тем распределение имеет выраженную правостороннюю асимметрию, высокий центральный пик и 
# длинный правый хвост с отдельными выбросами, выходящими за пределы +3 (вплоть до 6,6).Результаты статистических тестов также 
# подтверждают отклонение остатков от нормального распределения: для тестов Omnibus (434,254) и Jarque–Bera (2330,384) значение 
# p-value<0,05. Коэффициент асимметрии составляет 1,247, а эксцесса — 8,574, что существенным образом отличается от показателей 
# нормального распределения (которые должны быть равны нулю).


# Оценить значимость построенной модели
from scipy.stats import f

alpha = 0.05
df1 = 5  # количество включенных факторов (sqft_living_3, sqft_living_2_log, bedrooms_3, bedrooms_2_log, waterfront)
df2 = 1494  # n - df1 - 1 (1500 - 5 - 1)

F_crit = f.ppf(1 - alpha, df1, df2)

print('F_крит =', F_crit)

# Для оценки значимости уравнения регрессии сопоставим наблюдаемое значение F-статистики с критическим Fкрит:
# Значение из OLS-отчёта: F=354,5
# Критическое значение: Fкрит=2,220088 (при α=0,05, df1=5, df2=1494)
# Так как F>Fкрит(354,5>2,22), мы делаем вывод, что построенное уравнение регрессии в
#  целом является статистически значимым. Включённая совокупность факторов оказывает существенное и 
# неслучайное влияние на зависимую переменную price.
```

### Пуассоновская модель

Проверка сверхдисперсии и подготовка данных

```python
# Проверка пропусков
print(df.isnull().sum())

# Убедимся, что нет пропусков в выбранных столбцах
df = df.dropna(subset=['price', 'sqft_living', 'bedrooms', 'waterfront'])

# Среднее и дисперсия зависимой переменной
y = df['price']
print('Среднее price:', y.mean())
print('Дисперсия price:', y.var())
print('Минимум price:', y.min())
print('Максимум price:', y.max())
```

Построение квазипуассоновской регрессии

```python
import statsmodels.api as sm

# Формируем матрицу признаков
X = df[['sqft_living', 'bedrooms', 'waterfront']]
X = sm.add_constant(X)   # добавляем свободный член

# Квазипуассоновская модель (поправка на сверхдисперсию)
quasi_poisson_model = sm.GLM(
    y,
    X,
    family=sm.families.Poisson()
).fit(scale='X2')

print(quasi_poisson_model.summary())
```

График остатков Пирсона

```python
import matplotlib.pyplot as plt

fitted = quasi_poisson_model.fittedvalues
residuals = quasi_poisson_model.resid_pearson

plt.figure(figsize=(9, 6))
plt.scatter(fitted, residuals, alpha=0.6)
plt.axhline(0, linestyle='--', color='red')
plt.xlabel('Предсказанные значения цены')
plt.ylabel('Остатки Пирсона')
plt.title('График остатков квазипуассоновской модели')
plt.grid(True, alpha=0.3)
plt.show()
```

Гистограмма стандартизированных остатков

```python
influence = quasi_poisson_model.get_influence()
standardized_residuals = influence.resid_studentized

n = len(standardized_residuals)
k = int(np.ceil(1 + 3.322 * np.log10(n)))  # формула Стерджесса
xmin, xmax = standardized_residuals.min(), standardized_residuals.max()
bins = np.linspace(xmin, xmax, k + 1)

plt.figure(figsize=(9, 6))
plt.hist(standardized_residuals, bins=bins, edgecolor='black')
plt.xlabel('Стандартизированные остатки')
plt.ylabel('Частота')
plt.title('Гистограмма стандартизированных остатков квазипуассоновской модели')
plt.xticks(bins, [f'{x:.1f}' for x in bins], rotation=45)
plt.grid(True, alpha=0.3)
plt.show()
```
