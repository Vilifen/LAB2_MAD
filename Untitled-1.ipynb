```python
## ПУНКТЫ 1-3
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import f, chi2


df = pd.read_csv('ДанныеДляАнализа.csv', sep=';')
df.head()

# Из датасета для лабораторной работы выберем 4 переменные:
# price — цена объекта (зависимая количественная непрерывная переменная);
# sqft_living — жилая площадь в кв. футах (объясняющая количественная непрерывная переменная);
# bedrooms — количество спален (объясняющая количественная дискретная переменная);
# waterfront — вид на водоем (объясняющая качественная бинарная/фиктивная переменная: 0 — нет вида, 1 — есть вид).

# Удаление ненужных столбцов 
df = df[['price', 'sqft_living', 'bedrooms', 'waterfront']].copy()

print(df.shape)
df.head()

# Дескрептивный анализ для price 
p = df['price']

price_stats = {
    'Среднее':               p.mean(),
    'Стандартная ошибка':   p.sem(),
    'Медиана':              p.median(),
    'Мода':                 p.mode().iloc[0],
    'Стандартное отклонение': p.std(),
    'Дисперсия выборки':    p.var(),
    'Эксцесс':              p.kurt(),
    'Асимметричность':      p.skew(),
    'Интервал':             p.max() - p.min(),
    'Минимум':              p.min(),
    'Квартиль 25%':         p.quantile(0.25),
    'Квартиль 50%':         p.quantile(0.50),
    'Квартиль 75%':         p.quantile(0.75),
    'Максимум':             p.max(),
    'Сумма':                p.sum(),
    'Счет':                 p.count()
}

print(pd.Series(price_stats))

plt.figure(figsize=(8, 4))
plt.boxplot(p, vert=False)
plt.xlabel('Price')
plt.title('Ящик с усами для price')
plt.grid(axis='x', alpha=0.3)
plt.show()

# Гистограмма распределения price
price = df['price']
n = len(price)
k = int(np.ceil(1 + 3.322 * np.log10(n))) # k = 12
minimum = price.min()
maximum = price.max()
h = (maximum - minimum) / k

bins = np.linspace(minimum, maximum, k + 1)

plt.figure(figsize=(10, 6))
plt.hist(price, bins=bins, edgecolor='black')
plt.xticks(bins, [f'{x:.0f}' for x in bins], rotation=45)
plt.xlabel('Цена недвижимости (price)')
plt.ylabel('Частота')
plt.title('Гистограмма распределения переменной price')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Дескриптивный анализ для sqft_living
s = df['sqft_living']

sqft_stats = {
    'Среднее':               s.mean(),
    'Стандартная ошибка':   s.sem(),
    'Медиана':              s.median(),
    'Мода':                 s.mode().iloc[0],
    'Стандартное отклонение': s.std(),
    'Дисперсия выборки':    s.var(),
    'Эксцесс':              s.kurt(),
    'Асимметричность':      s.skew(),
    'Интервал':             s.max() - s.min(),
    'Минимум':              s.min(),
    'Квартиль 25%':         s.quantile(0.25),
    'Квартиль 50%':         s.quantile(0.50),
    'Квартиль 75%':         s.quantile(0.75),
    'Максимум':             s.max(),
    'Сумма':                s.sum(),
    'Счет':                 s.count()
}

print(pd.Series(sqft_stats))

plt.figure(figsize=(8, 4))
plt.boxplot(s, vert=False)
plt.xlabel('Sqft Living')
plt.title('Ящик с усами для sqft_living')
plt.grid(axis='x', alpha=0.3)
plt.show()


# Гистограмма sqft_living
n = len(s)
k = int(np.ceil(1 + 3.322 * np.log10(n)))
bins = np.linspace(s.min(), s.max(), k + 1)

plt.figure(figsize=(10, 5))
plt.hist(s, bins=bins, edgecolor='black')
plt.xlabel('Sqft Living')
plt.ylabel('Частота')
plt.title('Гистограмма распределения sqft_living')
plt.xticks(bins, rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()


# Дескриптивный анализ для bedrooms
b = df['bedrooms']

bedrooms_stats = {
    'Среднее':               b.mean(),
    'Стандартная ошибка':   b.sem(),
    'Медиана':              b.median(),
    'Мода':                 b.mode().iloc[0],
    'Стандартное отклонение': b.std(),
    'Дисперсия выборки':    b.var(),
    'Эксцесс':              b.kurt(),
    'Асимметричность':      b.skew(),
    'Интервал':             b.max() - b.min(),
    'Минимум':              b.min(),
    'Квартиль 25%':         b.quantile(0.25),
    'Квартиль 50%':         b.quantile(0.50),
    'Квартиль 75%':         b.quantile(0.75),
    'Максимум':             b.max(),
    'Сумма':                b.sum(),
    'Счет':                 b.count()
}

print(pd.Series(bedrooms_stats))

plt.figure(figsize=(8, 4))
plt.boxplot(b, vert=False)
plt.xlabel('Bedrooms')
plt.title('Ящик с усами для bedrooms')
plt.grid(axis='x', alpha=0.3)
plt.show()


# Дескриптивный анализ для переменной waterfront
wf = df['waterfront']
N = wf.count()

count_0 = (wf == 0).sum()
count_1 = (wf == 1).sum()

percent_0 = count_0 / N * 100 
percent_1 = count_1 / N * 100

mode_wf = wf.mode().iloc[0]

waterfront_stats = {
    'Мода':                             mode_wf,
    'Количество 0 (без вида на воду)': count_0,
    'Количество 1 (с видом на воду)':   count_1,
    'Процент 0':                        percent_0,
    'Процент 1':                        percent_1,
    'Счет':                             N
}

print(pd.Series(waterfront_stats))

# Столбчатая диаграмма
categories = ['Без вида (0)', 'С видом (1)']
counts = [count_0, count_1]

plt.figure(figsize=(6, 4))
plt.bar(categories, counts, edgecolor='black')
plt.ylabel('Частота')
plt.title('Распределение переменной waterfront')
for i, count in enumerate(counts):
    plt.text(i, count + 20, str(count), ha='center')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
